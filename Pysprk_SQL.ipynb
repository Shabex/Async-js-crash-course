## Defining Schemas in Pyspark

```python
schema = StructType ([
    StructField('user_id', IntegerType(), False),
    StructField('name', StringType(), True),
    StructField('gender', StringType(), True),
    StructField('age', IntegerType(), True),
    StructField('country', StringType(), True),
    StructField('email', StringType(), True),
    StructField('height', IntegerType(), True),
    StructField('sales', FloatType(), True)
])
```

## Viewing Your Data

```python
defined_df = spark.read.csv('users.csv', header=True, schema=schema)
```

```python
defined_df.show()
```

```python
defined_df.show(truncate = False)
```

```python
## Checking Columns

defined_df.columns
```

```python
defined_df.count()
```

```python
defined_df.select ('name','age','gender').show()
```

```python
defined_df.filter(defined_df.gender=='Male').show()
```

```python
defined_df.filter((defined_df.gender=='Male') & (defined_df.age < 35)).show()
```

```python
defined_df.where((defined_df.gender=='Female') & (defined_df.age >=40)).show()
```

```python
defined_df.show(5, truncate = False)
```

```python
defined_df = defined_df.withColumnRenamed('name','customer_name')
```

```python
defined_df.show(5, truncate=False)
```

```python
defined_df = defined_df.drop('height')
defined_df.show(5, truncate=False)
```

```python
defined_df.orderBy('age').show()
```

```python
defined_df.orderBy(F.col('age').desc()).show()
```

```python
defined_df = defined_df.withColumn('age_minus_5', F.col('age')-5)
defined_df.show()
```

```python
defined_df.groupBy('country').sum('sales').show()
```

```python
defined_df.groupBy('country').sum('age').show()
```

```python
defined_df.groupBy('gender').avg('age').show()
```

```python
defined_df.groupBy('country').agg(
    F.mean('sales').alias('average_sales'),
    F.max('age').alias('oldest_customer')
  ).show()
```

```python
defined_df.select(
    'country'
    F.round('sales',2).alias('rounded_sales')
).show()
```

```python
defined_df.select('country','sales').groupBy('country').agg(F.sum('sales' ).alias('total_sales')
).show()
```

## SPARK SQL
*   It is a component of Apache Sparch for querying structured data using SQL Syntax.
*  It works with DataFrames and Datasets.
*  Key Benefit : Combines SQL familiarity with spark scalability.

```python
## Register a Temp View
## This lets you query it with SQL:

defined_df.createOrReplaceTempView('employees')
```

```python
# Show all
spark.sql('select * from employees').show()
```

```python
# Filtering
spark.sql('select * from employees where sales > 1000000').show()
```

```python
# Filtering
spark.sql('select * from employees where sales > 1000000 order by sales Desc limit 1').show()
```
